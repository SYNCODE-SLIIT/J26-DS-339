## Interpretation

Candidate agreement is the proportion of common voiced interior frames within 50 cents of a shifted MIDI reference. These are selected problematic TRAIN cases, not validation accuracy. A favorable shifted hypothesis is not proof of the performed fundamental, and imperfect agreement may involve timing, intonation, melody differences or estimator errors.
